```python
%%writefile distributed_cuda_solver.py

import os
import sys
import json
import math
import hashlib
import subprocess
import numpy as np

from mpi4py import MPI
from numba import cuda, float64

TILE_DIM = 16
SHARED_DIM = TILE_DIM + 2

GLOBAL_NX = 512
GLOBAL_NY = 512
STEPS = 400
LAMBDA_DAMP = 0.15

def get_gpu_uuid() -> str:
    """Get GPU UUID from NVIDIA GPU."""
    out = subprocess.check_output(
        ["nvidia-smi", "--query-gpu=uuid", "--format=csv,noheader"],
        encoding="utf-8",
    )
    return out.strip().splitlines()[0].strip()


def derive_student_seed(github_user: str, gpu_uuid: str) -> int:
    """Generate unique seed using GitHub username and GPU UUID."""
    digest = hashlib.sha256(
        f"{github_user.strip()}::{gpu_uuid}".encode("utf-8")
    ).hexdigest()

    return int(digest[:16], 16)


def init_distributed_context(github_user: str):
    comm = MPI.COMM_WORLD
    rank = comm.Get_rank()
    size = comm.Get_size()

    if size != 4:
        raise RuntimeError(
            f"Expected exactly 4 MPI ranks, got {size}."
        )

    cuda.select_device(0)

    gpu_uuid = get_gpu_uuid() if rank == 0 else None
    gpu_uuid = comm.bcast(gpu_uuid, root=0)

    instance_seed = derive_student_seed(github_user, gpu_uuid)

    dims = [2, 2]
    periods = [False, False]

    cart_comm = comm.Create_cart(
        dims=dims,
        periods=periods,
        reorder=False
    )

    coords = cart_comm.Get_coords(rank)

    nbr_up, nbr_down = cart_comm.Shift(0, 1)
    nbr_left, nbr_right = cart_comm.Shift(1, 1)

    print(
        f"Rank {rank}: coords={coords}, "
        f"up={nbr_up}, down={nbr_down}, "
        f"left={nbr_left}, right={nbr_right}",
        flush=True
    )

    return (
        cart_comm,
        rank,
        coords,
        nbr_up,
        nbr_down,
        nbr_left,
        nbr_right,
        instance_seed,
    )

    # ============================================================
# TASK 2: SEEDED HETEROGENEOUS SUBDOMAIN INITIALIZATION
# ============================================================

def initialize_subdomains(coords, instance_seed):
    ny = GLOBAL_NY // 2
    nx = GLOBAL_NX // 2

    dx = 1.0 / (GLOBAL_NX + 1)
    dy = 1.0 / (GLOBAL_NY + 1)

    row_rank, col_rank = coords

    # Local indices, including ghost cells
    local_rows = np.arange(ny + 2)
    local_cols = np.arange(nx + 2)

    # Convert local indices to global physical coordinates
    global_rows = row_rank * ny + local_rows
    global_cols = col_rank * nx + local_cols

    y = global_rows * dy
    x = global_cols * dx

    X, Y = np.meshgrid(x, y)

    # Heterogeneous diffusivity in [0.08, 0.24]
    kappa = (
        0.16
        + 0.04 * np.sin(2.0 * np.pi * X)
        + 0.04 * np.cos(2.0 * np.pi * Y)
    )

    # Seed-dependent Gaussian pulse parameters
    rng = np.random.default_rng(instance_seed)

    center_x = 0.3 + 0.4 * rng.random()
    center_y = 0.3 + 0.4 * rng.random()
    amplitude = 0.8 + 0.4 * rng.random()

    sigma = 0.08

    u = amplitude * np.exp(
        -(
            (X - center_x) ** 2
            + (Y - center_y) ** 2
        ) / (2.0 * sigma ** 2)
    )

    # Localized source field
    source = 0.02 * np.exp(
        -(
            (X - 0.5) ** 2
            + (Y - 0.5) ** 2
        ) / (2.0 * 0.05 ** 2)
    )

    # Physical boundary conditions
    if row_rank == 0:
        u[0, :] = 0.0

    if row_rank == 1:
        u[-1, :] = 0.0

    if col_rank == 0:
        u[:, 0] = 0.0

    if col_rank == 1:
        u[:, -1] = 0.0

    # Transfer local arrays to GPU
    d_u = cuda.to_device(u.astype(np.float64))
    d_kappa = cuda.to_device(kappa.astype(np.float64))
    d_source = cuda.to_device(source.astype(np.float64))

    return d_u, d_kappa, d_source, dx, dy


    # ============================================================
# TASK 3: GPU HALO PACK / UNPACK KERNELS
# ============================================================

@cuda.jit
def pack_halos_kernel(u, top, bottom, left, right, ny, nx):
    idx = cuda.grid(1)

    if idx < nx:
        top[idx] = u[1, idx + 1]
        bottom[idx] = u[ny, idx + 1]

    if idx < ny:
        left[idx] = u[idx + 1, 1]
        right[idx] = u[idx + 1, nx]


@cuda.jit
def unpack_halos_kernel(u, top, bottom, left, right, ny, nx,
                        has_top, has_bottom, has_left, has_right):
    idx = cuda.grid(1)

    if idx < nx:
        if has_top:
            u[0, idx + 1] = top[idx]
        else:
            u[0, idx + 1] = 0.0

        if has_bottom:
            u[ny + 1, idx + 1] = bottom[idx]
        else:
            u[ny + 1, idx + 1] = 0.0

    if idx < ny:
        if has_left:
            u[idx + 1, 0] = left[idx]
        else:
            u[idx + 1, 0] = 0.0

        if has_right:
            u[idx + 1, nx + 1] = right[idx]
        else:
            u[idx + 1, nx + 1] = 0.0


class HaloBuffers:
    def __init__(self, ny, nx):
        self.d_send = {
            "top": cuda.device_array(nx, dtype=np.float64),
            "bottom": cuda.device_array(nx, dtype=np.float64),
            "left": cuda.device_array(ny, dtype=np.float64),
            "right": cuda.device_array(ny, dtype=np.float64),
        }

        self.d_recv = {
            name: cuda.device_array(arr.shape, dtype=np.float64)
            for name, arr in self.d_send.items()
        }

        self.h_send = {
            name: cuda.pinned_array(arr.shape, dtype=np.float64)
            for name, arr in self.d_send.items()
        }

        self.h_recv = {
            name: cuda.pinned_array(arr.shape, dtype=np.float64)
            for name, arr in self.d_send.items()
        }


def exchange_halos_async(
    d_u, buffers, cart_comm, neighbors, stream_boundary
):
    ny = d_u.shape[0] - 2
    nx = d_u.shape[1] - 2

    threads = 256
    blocks = (max(nx, ny) + threads - 1) // threads

    pack_halos_kernel[blocks, threads, stream_boundary](
        d_u,
        buffers.d_send["top"],
        buffers.d_send["bottom"],
        buffers.d_send["left"],
        buffers.d_send["right"],
        ny, nx
    )

    for name in buffers.d_send:
        buffers.d_send[name].copy_to_host(
            buffers.h_send[name],
            stream=stream_boundary
        )

    # Ensure the host send buffers contain completed DMA data.
    stream_boundary.synchronize()

    opposite = {
        "top": "bottom",
        "bottom": "top",
        "left": "right",
        "right": "left",
    }

    tags = {
        "top": 10,
        "bottom": 11,
        "left": 12,
        "right": 13,
    }

    requests = []

    # Post receives first.
    for direction, neighbor in neighbors.items():
        requests.append(
            cart_comm.Irecv(
                buffers.h_recv[direction],
                source=neighbor,
                tag=tags[opposite[direction]]
            )
        )

    # Then post sends.
    for direction, neighbor in neighbors.items():
        requests.append(
            cart_comm.Isend(
                buffers.h_send[direction],
                dest=neighbor,
                tag=tags[direction]
            )
        )

    return requests


def finish_halo_exchange(
    d_u, buffers, requests, neighbors, stream_boundary
):
    MPI.Request.Waitall(requests)

    for name, neighbor in neighbors.items():
        if neighbor != MPI.PROC_NULL:
            buffers.d_recv[name].copy_to_device(
                buffers.h_recv[name],
                stream=stream_boundary
            )

    ny = d_u.shape[0] - 2
    nx = d_u.shape[1] - 2

    threads = 256
    blocks = (max(nx, ny) + threads - 1) // threads

    unpack_halos_kernel[blocks, threads, stream_boundary](
        d_u,
        buffers.d_recv["top"],
        buffers.d_recv["bottom"],
        buffers.d_recv["left"],
        buffers.d_recv["right"],
        ny, nx,
        neighbors["top"] != MPI.PROC_NULL,
        neighbors["bottom"] != MPI.PROC_NULL,
        neighbors["left"] != MPI.PROC_NULL,
        neighbors["right"] != MPI.PROC_NULL
    )

    stream_boundary.synchronize()


    # ============================================================
# TASK 4: SHARED-MEMORY TILED STENCIL
# ============================================================

@cuda.jit
def tiled_stencil_kernel(
    u, u_new, kappa, source,
    dx, dy, dt,
    r_start, r_end, c_start, c_end
):
    tx = cuda.threadIdx.x
    ty = cuda.threadIdx.y

    bx = cuda.blockIdx.x
    by = cuda.blockIdx.y

    # Global local-array coordinates
    j = bx * TILE_DIM + tx + 1
    i = by * TILE_DIM + ty + 1

    # Shared memory: 16x16 interior + halo
    smem = cuda.shared.array(
        shape=(SHARED_DIM, SHARED_DIM),
        dtype=float64
    )

    sy = ty + 1
    sx = tx + 1

    ny = u.shape[0] - 2
    nx = u.shape[1] - 2

    # Load center values
    if i <= ny and j <= nx:
        smem[sy, sx] = u[i, j]
    else:
        smem[sy, sx] = 0.0

    # Load halo apron
    if tx == 0 and i <= ny:
        smem[sy, 0] = u[i, j - 1]

    if tx == TILE_DIM - 1 and i <= ny and j <= nx:
        smem[sy, SHARED_DIM - 1] = u[i, j + 1]

    if ty == 0 and j <= nx:
        smem[0, sx] = u[i - 1, j]

    if ty == TILE_DIM - 1 and i <= ny and j <= nx:
        smem[SHARED_DIM - 1, sx] = u[i + 1, j]

    cuda.syncthreads()

    # Bounding-box restriction
    if (
        i >= r_start and i <= r_end
        and j >= c_start and j <= c_end
        and i <= ny and j <= nx
    ):
        center = smem[sy, sx]

        up = smem[sy - 1, sx]
        down = smem[sy + 1, sx]
        left = smem[sy, sx - 1]
        right = smem[sy, sx + 1]

        kc = kappa[i, j]

        kup = 0.5 * (kc + kappa[i - 1, j])
        kdown = 0.5 * (kc + kappa[i + 1, j])
        kleft = 0.5 * (kc + kappa[i, j - 1])
        kright = 0.5 * (kc + kappa[i, j + 1])

        diffusion_x = (
            kright * (right - center)
            - kleft * (center - left)
        ) / (dx * dx)

        diffusion_y = (
            kdown * (down - center)
            - kup * (center - up)
        ) / (dy * dy)

        reaction = -LAMBDA_DAMP * center ** 3

        u_new[i, j] = center + dt * (
            diffusion_x
            + diffusion_y
            + reaction
            + source[i, j]
        )

        # ============================================================
# TASK 4: DUAL-STREAM COMPUTE-COMMUNICATION OVERLAP
# ============================================================

def step_overlapped(
    d_u, d_u_new, d_kappa, d_source,
    dx, dy, dt,
    buffers, cart_comm, neighbors,
    stream_boundary, stream_interior
):
    ny = d_u.shape[0] - 2
    nx = d_u.shape[1] - 2

    threads = (TILE_DIM, TILE_DIM)
    blocks = (
        (nx + TILE_DIM - 1) // TILE_DIM,
        (ny + TILE_DIM - 1) // TILE_DIM
    )

    # Stage A + B: pack halos and post non-blocking MPI transfers
    requests = exchange_halos_async(
        d_u, buffers, cart_comm, neighbors, stream_boundary
    )

    # Deep interior does not depend on incoming halos
    tiled_stencil_kernel[
        blocks, threads, stream_interior
    ](
        d_u, d_u_new, d_kappa, d_source,
        dx, dy, dt,
        2, ny - 1, 2, nx - 1
    )

    # Stage C: finish MPI transfers and unpack ghost cells
    finish_halo_exchange(
        d_u, buffers, requests, neighbors, stream_boundary
    )

    # Update top boundary
    tiled_stencil_kernel[
        blocks, threads, stream_boundary
    ](
        d_u, d_u_new, d_kappa, d_source,
        dx, dy, dt,
        1, 1, 1, nx
    )

    # Update bottom boundary
    tiled_stencil_kernel[
        blocks, threads, stream_boundary
    ](
        d_u, d_u_new, d_kappa, d_source,
        dx, dy, dt,
        ny, ny, 1, nx
    )

    # Update left boundary
    tiled_stencil_kernel[
        blocks, threads, stream_boundary
    ](
        d_u, d_u_new, d_kappa, d_source,
        dx, dy, dt,
        2, ny - 1, 1, 1
    )

    # Update right boundary
    tiled_stencil_kernel[
        blocks, threads, stream_boundary
    ](
        d_u, d_u_new, d_kappa, d_source,
        dx, dy, dt,
        2, ny - 1, nx, nx
    )

    # Both streams must finish before swapping arrays
    stream_interior.synchronize()
    stream_boundary.synchronize()

    return d_u_new, d_u

    # ============================================================
# TASK 5: SERIALIZED SOLVER
# ============================================================

def step_serialized(
    d_u, d_u_new, d_kappa, d_source,
    dx, dy, dt,
    buffers, cart_comm, neighbors,
    stream_boundary
):
    ny = d_u.shape[0] - 2
    nx = d_u.shape[1] - 2

    threads = (TILE_DIM, TILE_DIM)

    blocks = (
        (nx + TILE_DIM - 1) // TILE_DIM,
        (ny + TILE_DIM - 1) // TILE_DIM
    )

    # Complete halo exchange before computation
    requests = exchange_halos_async(
        d_u, buffers, cart_comm, neighbors, stream_boundary
    )

    finish_halo_exchange(
        d_u, buffers, requests, neighbors, stream_boundary
    )

    # Compute the entire owned subdomain
    tiled_stencil_kernel[
        blocks, threads, stream_boundary
    ](
        d_u, d_u_new, d_kappa, d_source,
        dx, dy, dt,
        1, ny, 1, nx
    )

    stream_boundary.synchronize()

    return d_u_new, d_u


    # ============================================================
# TASK 5: GLOBAL L2 RESIDUAL
# ============================================================

def compute_global_l2_residual(d_u, d_u_previous, cart_comm):
    current = d_u.copy_to_host()[1:-1, 1:-1]
    previous = d_u_previous.copy_to_host()[1:-1, 1:-1]

    local_squared = np.array(
        [np.sum((current - previous) ** 2)],
        dtype=np.float64
    )

    global_squared = np.zeros(1, dtype=np.float64)

    cart_comm.Allreduce(
        local_squared,
        global_squared,
        op=MPI.SUM
    )

    return np.sqrt(global_squared[0])
```

### Question 1: Memory Coalescing & Strided Column Packing

In my project, I used a 512×512 grid divided between 4 MPI ranks. Each rank works with a 256×256 part of the grid.

I used a CUDA kernel to pack the halo data before sending it to another rank. The top and bottom rows are stored next to each other in memory, so GPU threads can read them efficiently.

The left and right columns are different because their elements are separated by the row length, which is 258 elements including the ghost cells. This makes column access less efficient.

To solve this, I used a CUDA packing kernel to copy the column values into continuous buffers. After that, MPI can send the data as a normal array.

This method makes communication easier, but the GPU still needs strided memory reads when packing columns.

### Question 2: Shared-Memory Tile Indexing & Bank Conflict Analysis

I used shared memory to reduce the number of global memory accesses in my CUDA stencil.

My CUDA block has 16×16 threads, and the shared-memory tile is 18×18. The extra cells are needed to store the neighboring values around the block.

Each thread loads its main value into shared memory. Threads at the edges also load halo values.

For example, thread `(0, 0)` stores its main value at shared-memory position `(1, 1)`. The extra positions are used for the neighboring cells.

After loading the values, I used `cuda.syncthreads()` to make sure all threads finished loading before the calculation started.

The shared-memory array uses double-precision values. On the Tesla T4, memory-bank conflicts can happen because of the access pattern and the size of each value.

The 18-column layout does not completely remove bank conflicts. However, shared memory helps the threads reuse neighboring values instead of reading everything from global memory again.

### Question 3: PTX Assembly Audit & Warp Synchronization

I checked the PTX assembly generated by my CUDA stencil kernel.

My results were:

- `st.shared.f64`: 5
- `ld.shared.f64`: 5
- `bar.sync`: 1

The `st.shared.f64` instructions show that the kernel writes double-precision values into shared memory.

The `ld.shared.f64` instructions show that it reads values from shared memory during the stencil calculation.

The `bar.sync` instruction is important because the threads must finish loading the tile before using the neighboring values.

This is a block-level synchronization barrier, not only a warp-level barrier.

My PTX audit passed because all three instruction types were present in the generated code.

### Question 4: Multi-Process Context Sharing on a Single Physical GPU

I used Google Colab with one NVIDIA Tesla T4 GPU.

My MPI program runs with 4 ranks, but all 4 ranks use the same physical GPU.

Each MPI rank creates its own CUDA context and works with its own part of the grid. Because they share one GPU, the processes compete for GPU resources.

I tested two execution methods.

My first 400-step benchmark results were:

- Serialized: 5336.785 ms
- Overlapped: 1861.280 ms

The overlapped version was about 2.87 times faster in this run.

However, these results can be affected by CUDA compilation and other overhead.

I also measured 10 steps using CUDA Events:

- Serialized: 45.525 ms
- Overlapped: 60.885 ms

In this shorter test, the serialized version had a lower measured time.

I think this difference shows why performance testing is difficult when several MPI processes share one GPU. The results depend on what we measure and how the GPU schedules the work.

### Question 5: Point-to-Point Message Deadlock & MPI.PROC_NULL Semantics

In my project, I created a 2×2 MPI Cartesian topology.

Each rank communicates with its neighboring ranks to exchange halo values.

I used `MPI.Irecv()` and `MPI.Isend()` for nonblocking communication. First, I posted the receive operations, and then I posted the send operations.

After that, I used `MPI.Request.Waitall()` to wait until communication was completed.

This approach helps avoid the deadlock problems that can happen when processes use blocking sends and wait for each other.

I also used `MPI.PROC_NULL` for boundaries without a neighboring rank.

For example, rank 0 has no top or left neighbor. In my output, these neighbors were shown as `-2`, which is the value of `MPI.PROC_NULL` in my environment.

Communication with `MPI.PROC_NULL` completes without transferring data.

For these physical boundaries, my CUDA unpacking kernel sets the ghost values to zero.

This allows the same communication code to work for all four ranks.
